```python
source_path = "/Volumes/workspace/greenhouse_bronze/raw_files/32100456.csv"

files = dbutils.fs.ls(
    "/Volumes/workspace/greenhouse_bronze/raw_files/"
)

display(files)



```

```python
# Read the original Statistics Canada CSV into a Spark DataFrame

df_raw = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "false")
    .option("encoding", "UTF-8")
    .csv(source_path)
)

# Inspect the structure and first 10 rows
df_raw.printSchema()
display(df_raw.limit(10))

# Count the records
print("Total source records:", df_raw.count())
```

```python

from pyspark.sql.functions import current_timestamp, lit

# Add ingestion metadata without changing original source columns
df_bronze = (
    df_raw
    .withColumn("ingestion_timestamp", current_timestamp())
    .withColumn("source_file", lit(source_path))
)

# Write Bronze Delta table while preserving original column names
(
    df_bronze.write
    .format("delta")
    .mode("overwrite")
    .option("delta.columnMapping.mode", "name")
    .saveAsTable("workspace.greenhouse_bronze.raw_production")
)

# Verify record count
bronze_count = spark.table(
    "workspace.greenhouse_bronze.raw_production"
).count()

print("Source records:", df_raw.count())
print("Bronze records:", bronze_count)
print("Counts match:", df_raw.count() == bronze_count)

```

```python

# Read the Bronze Delta table
df_check = spark.table(
    "workspace.greenhouse_bronze.raw_production"
)

# Inspect the schema
df_check.printSchema()

# Check source metadata
display(
    df_check.select(
        "REF_DATE",
        "GEO",
        "Production and value",
        "VALUE",
        "ingestion_timestamp",
        "source_file"
    ).limit(10)
)

# Check for missing ingestion metadata
from pyspark.sql.functions import col

print(
    "Missing source paths:",
    df_check.filter(col("source_file").isNull()).count()
)

print(
    "Missing ingestion timestamps:",
    df_check.filter(col("ingestion_timestamp").isNull()).count()
)
```
